```python
def calculate_threshold(points) :
        i=0
        sum=0
        size=len(points)
        for i in range(size-1) :
                sum=sum+abs(points[i][0]-points[i+1][0])
        return abs(sum)/size

def calculate_curvature(points):
    """
    Calculate the curvature at each point in a series of points.
    Parameters:
    - points: Nx2 array of x and y coordinates.
    Returns:
    - curvature: N array of curvature values.
    """
    # def calculate_curvature(x, y):
    # dy = np.gradient(y, x)
    # d2y = np.gradient(dy, x)
    # curvature = np.abs(d2y)
    # return curvature

    dx = np.gradient(points[:, 1])
    dy = np.gradient(points[:, 0])
    ddx = np.gradient(dx)
    ddy = np.gradient(dy)
    numerator = dx * ddy - dy * ddx
    denominator = (dx**2 + dy**2)**1.5
    # Avoid division by zero
    denominator[denominator == 0] = np.finfo(float).eps
    curvature = np.abs(numerator / denominator)
    return curvature

def curvature_rdp(points, epsilon, curvature_threshold):
    """
    Reduce the number of points in a curve using a curvature-aware RDP algorithm.
    Parameters:
    - points: Nx2 array of x and y coordinates.
    - epsilon: Tolerance parameter for distance.
    - curvature_threshold: Tolerance parameter for curvature.
    Returns:
    - Reduced set of points.
    """
    if len(points) < 3:
        return points

    # Calculate curvature
    curvature = calculate_curvature(points)

    # Find the point with the maximum combined metric (distance * curvature)
    start, end = points[0], points[-1]
    line_vec = end - start
    line_len = np.hypot(*line_vec)
    if line_len == 0:
        line_len = np.finfo(float).eps  # Avoid division by zero
    line_vec_norm = line_vec / line_len
    vec_from_start = points - start
    proj_lengths = np.dot(vec_from_start, line_vec_norm)
    proj_points = start + np.outer(proj_lengths, line_vec_norm)
    distances = np.linalg.norm(points - proj_points, axis=1)
    
    # Combine distance and curvature
    combined_metric = distances * curvature

    max_metric = np.max(combined_metric)
    index = np.argmax(combined_metric)

    # If the max metric is greater than thresholds, recursively simplify
    if max_metric > epsilon or curvature[index] > curvature_threshold:
        left = curvature_rdp(points[:index+1], epsilon, curvature_threshold)
        right = curvature_rdp(points[index:], epsilon, curvature_threshold)
        return np.vstack((left[:-1], right))
    else:
        return np.array([start, end])

def curvature_rdp_limited(points, max_nodes, initial_epsilon, initial_curvature_threshold, step):
    epsilon = initial_epsilon
    curvature_threshold = initial_curvature_threshold
    while True:
        reduced_points = curvature_rdp(points, epsilon, curvature_threshold)
        if len(reduced_points) <= max_nodes:
            return reduced_points
        else:
            # Simplify more by increasing epsilon and curvature_threshold
            epsilon += step
            curvature_threshold += step

# Main algorithm
for i in range(100, 110):
    dataX = np.array(localXe[i])
    dataY = np.array(localYe[i])
    points = np.column_stack((dataX, dataY))
    max_nodes = 7
    epsilon = calculate_threshold(points) *2
    curvature_threshold = np.median(calculate_curvature(points)) / 5
    step = epsilon / 5

    reduced_points = curvature_rdp_limited(points, max_nodes, epsilon, curvature_threshold, step)
    x_nodes = reduced_points[:, 0]
    y_nodes = reduced_points[:, 1]

    plt.figure(figsize=(20, 10))
    plt.title(f"Track {allPedIds[i]}")
    plt.plot(dataX, dataY, label='Original Data', color='blue')
    plt.plot(x_nodes, y_nodes, label='Curvature-Prioritized Spline', color='orange')
    plt.scatter(x_nodes, y_nodes, color='red', label='Spline Knots')
    plt.legend()
    plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.preprocessing import MinMaxScaler

def point_line_distance(point, start, end):
    """Calculate perpendicular distance between a point and a line segment"""
    # if start == end:
    if np.array_equal(start, end):
        return ((point[0] - start[0]) ** 2 + (point[1] - start[1]) ** 2) ** 0.5
    
    n = abs(
        (end[0] - start[0]) * (start[1] - point[1]) - 
        (start[0] - point[0]) * (end[1] - start[1])
    )
    d = ((end[0] - start[0]) ** 2 + (end[1] - start[1]) ** 2) ** 0.5
    
    return n/d

def douglas_peucker(points, epsilon):
    """
    Reduce number of points in a curve using Douglas-Peucker algorithm
    
    Args:
        points: List of (x,y) tuples
        epsilon: Maximum distance for point elimination
        
    Returns:
        List of (x,y) tuples after simplification
    """
    if len(points) <= 2:
        return points
        
    d_max = 0
    index = 0
    
    for i in range(1, len(points) - 1):
        d = point_line_distance(points[i], points[0], points[-1])
        if d > d_max:
            index = i
            d_max = d
            
    if d_max > epsilon:
        rec_results1 = douglas_peucker(points[:index + 1], epsilon)
        rec_results2 = douglas_peucker(points[index:], epsilon)
        return rec_results1[:-1] + rec_results2
    else:
        return [points[0], points[-1]]

def calculate_dynamic_epsilon(points, method='stats', target_reduction=0.5):
    """
    Calculate epsilon dynamically based on different methods
    
    Args:
        points: List of (x,y) tuples
        method: String indicating the method to use
            'stats' - based on statistical properties
            'range' - based on data range
            'adaptive' - uses point-to-point distances
            'percentile' - uses distance percentiles
        target_reduction: Target fraction of points to keep (0-1)
            
    Returns:
        float: calculated epsilon value
    """
    x, y = zip(*points)
    
    if method == 'stats':
        # Based on standard deviation of distances
        distances = [point_line_distance(points[i], points[0], points[-1])
                    for i in range(1, len(points)-1)]
        epsilon = np.std(distances) * (1 - target_reduction)
        
    elif method == 'range':
        # Based on data range
        x_range = max(x) - min(x)
        y_range = max(y) - min(y)
        epsilon = min(x_range, y_range) * 0.01 * (1 - target_reduction)
        
    elif method == 'adaptive':
        # Based on average point-to-point distances
        distances = []
        for i in range(len(points)-1):
            dist = np.sqrt((points[i+1][0] - points[i][0])**2 + (points[i+1][1] - points[i][1])**2)
            distances.append(dist)
        epsilon = np.mean(distances) * (1 - target_reduction)
        
    elif method == 'percentile':
        # Based on distance percentiles
        distances = [point_line_distance(points[i], points[0], points[-1])
                    for i in range(1, len(points)-1)]
        epsilon = np.percentile(distances, target_reduction * 100)
        
    else:
        raise ValueError("Unknown method")
        
    return epsilon

def simplify_with_analysis(points, method='stats', target_reduction=0.5):
    """
    Simplify points and provide analysis
    
    Args:
        points: List of (x,y) tuples
        method: Method for epsilon calculation
        target_reduction: Target fraction of points to keep
        
    Returns:
        dict: Results including simplified points and metrics
    """
    epsilon = calculate_dynamic_epsilon(points, method, target_reduction)
    simplified_points = douglas_peucker(points, epsilon)
    
    # Calculate error metrics
    original_length = len(points)
    simplified_length = len(simplified_points)
    reduction_ratio = (original_length - simplified_length) / original_length
    
    # Calculate approximation error
    error = 0
    for point in points:
        min_dist = float('inf')
        for i in range(len(simplified_points)-1):
            dist = point_line_distance(point, simplified_points[i], simplified_points[i+1])
            min_dist = min(min_dist, dist)
        error += min_dist
    avg_error = error / len(points)
    
    return {
        'simplified_points': simplified_points,
        'epsilon': epsilon,
        'original_points': len(points),
        'simplified_points_count': len(simplified_points),
        'reduction_ratio': reduction_ratio,
        'average_error': avg_error
    }

def plot_results(points, results_dict):
    """Plot original and simplified curves with metrics"""
    plt.figure(figsize=(15, 8))
    
    # Plot data
    x, y = zip(*points)
    plt.plot(x, y, label='Original')
    
    sx, sy = zip(*results_dict['simplified_points'])
    plt.plot(sx, sy, 'r.-', label='Simplified', linewidth=2)
    
    # Add metrics text
    metrics_text = (
        f"Epsilon: {results_dict['epsilon']:.4f}\n"
        f"Original points: {results_dict['original_points']}\n"
        f"Simplified points: {results_dict['simplified_points_count']}\n"
        f"Reduction ratio: {results_dict['reduction_ratio']:.2%}\n"
        f"Average error: {results_dict['average_error']:.4f}"
    )
    plt.text(0.02, 0.98, metrics_text, transform=plt.gca().transAxes, 
             verticalalignment='top', bbox=dict(boxstyle='round', facecolor='white', alpha=0.8))
    
    plt.grid(True)
    plt.legend()
    plt.title('Douglas-Peucker Simplification with Dynamic Epsilon')
    plt.show()

# Example usage
def test_methods(points):
    methods = ['stats', 'adaptive', 'percentile']
    target_reductions = [0.002, 0.2, 0.3, 0.5, 0.7]
    
    for method in methods:
        for target in target_reductions:
            print(f"\nTesting {method} method with target reduction {target}")
            results = simplify_with_analysis(points, method=method, target_reduction=target)
            plot_results(points, results)

# Generate sample data
# x = np.linspace(0, 4*np.pi, 200)
# y = np.sin(x) + np.random.normal(0, 0.1, 200)
# points = list(zip(x, y))

# # Test all methods
# test_methods(points)
for i in range(11 ,12):
    dataX = np.array(localXe[i])
    dataY = np.array(localYe[i])
    points = list(zip(dataX, dataY))
    test_methods(points)
```

```python
import numpy as np
from scipy.optimize import minimize



def optimize_simplified_path(original_points, simplified_points, max_points=None):
    """
    Optimize a simplified path by adjusting point positions and potentially adding/removing points.
    
    Args:
        original_points: List of (x,y) tuples of the original path
        simplified_points: Initial simplified points from RDP
        max_points: Maximum number of points allowed (optional)
    """
    def total_error(points_flat):
        # Reshape flat array back to points
        points = points_flat.reshape(-1, 2)
        
        # Calculate error between original and simplified
        total_err = 0
        for orig_point in original_points:
            # Find minimum distance to any segment
            min_dist = float('inf')
            for i in range(len(points) - 1):
                dist = point_line_distance(orig_point, 
                                        (points[i][0], points[i][1]),
                                        (points[i+1][0], points[i+1][1]))
                min_dist = min(min_dist, dist)
            total_err += min_dist
            
        # Add regularization to prevent points from moving too far
        reg_term = 0.1 * np.sum(np.abs(points_flat - np.array(simplified_points).flatten()))
        
        return total_err + reg_term

    # Convert points to flat array for optimization
    initial_guess = np.array(simplified_points).flatten()
    
    # Optimize point positions
    result = minimize(total_error, initial_guess, method='Powell')
    optimized_points = result.x.reshape(-1, 2)

    # Now consider adding/removing points
    while True:
        current_error = total_error(optimized_points.flatten())
        
        # Try adding a point where error is highest
        worst_error = 0
        best_new_point = None
        
        for orig_point in original_points:
            # Find segment with highest error
            for i in range(len(optimized_points) - 1):
                dist = point_line_distance(orig_point,
                                        (optimized_points[i][0], optimized_points[i][1]),
                                        (optimized_points[i+1][0], optimized_points[i+1][1]))
                if dist > worst_error:
                    worst_error = dist
                    best_new_point = orig_point
                    insert_idx = i + 1

        if best_new_point and (max_points is None or len(optimized_points) < max_points):
            # Try inserting the new point
            new_points = np.insert(optimized_points, insert_idx, best_new_point, axis=0)
            new_error = total_error(new_points.flatten())
            
            if new_error < current_error * 0.95:  # 5% improvement threshold
                optimized_points = new_points
                continue

        # Try removing each point and see if error doesn't increase much
        best_removal = None
        min_removal_error = float('inf')
        
        for i in range(1, len(optimized_points) - 1):
            test_points = np.delete(optimized_points, i, axis=0)
            removal_error = total_error(test_points.flatten())
            
            if removal_error < min_removal_error and removal_error < current_error * 1.05:  # 5% tolerance
                min_removal_error = removal_error
                best_removal = i

        if best_removal is not None:
            optimized_points = np.delete(optimized_points, best_removal, axis=0)
            continue
            
        break  # No more improvements possible

    return optimized_points.tolist()
```

```python
def calculate_error(original_points, simplified_points):
    """
    Calculate average and maximum error between original and simplified curves
    
    Args:
        original_points: List of (x,y) tuples for original curve
        simplified_points: List of (x,y) tuples for simplified curve
    
    Returns:
        dict: Error metrics
    """
    total_error = 0
    max_error = 0
    
    for orig_point in original_points:
        # Find minimum distance to any segment in simplified curve
        min_dist = float('inf')
        for i in range(len(simplified_points) - 1):
            dist = point_line_distance(orig_point,
                                     simplified_points[i],
                                     simplified_points[i+1])
            min_dist = min(min_dist, dist)
        
        total_error += min_dist
        max_error = max(max_error, min_dist)
    
    return {
        'average_error': total_error / len(original_points),
        'max_error': max_error
    }

def calculate_smoothness(points):
    """
    Calculate smoothness metric based on angle changes between segments
    
    Args:
        points: List of (x,y) tuples
    
    Returns:
        float: Smoothness metric (lower is smoother)
    """
    if len(points) < 3:
        return 0
    
    # Calculate angles between consecutive segments
    angles = []
    for i in range(len(points) - 2):
        p1 = np.array(points[i])
        p2 = np.array(points[i+1])
        p3 = np.array(points[i+2])
        
        v1 = p2 - p1
        v2 = p3 - p2
        
        # Calculate angle between vectors
        cos_angle = np.dot(v1, v2) / (np.linalg.norm(v1) * np.linalg.norm(v2))
        # Clip to avoid numerical errors
        cos_angle = np.clip(cos_angle, -1.0, 1.0)
        angle = np.arccos(cos_angle)
        angles.append(angle)
    
    # Return standard deviation of angles as smoothness metric
    return np.std(angles)

def analyze_parameter_sensitivity(points, param_ranges):
    """
    Analyze how different parameters affect the simplification
    
    Args:
        points: Original points
        param_ranges: Dict of parameters and their test values
    """
    results = {}
    
    for method in param_ranges['methods']:
        for target in param_ranges['target_reductions']:
            for reg in param_ranges['reg_strength']:
                print(f"Testing method={method}, target={target}, reg={reg}")
                key = (method, target, reg)
                
                # Run simplification
                epsilon = calculate_dynamic_epsilon(points, method, target)
                initial = douglas_peucker(points, epsilon)
                optimized = optimize_simplified_path(points, initial, 
                                                  reg_strength=reg)
                
                # Calculate metrics
                error_metrics = calculate_error(points, optimized)
                results[key] = {
                    'points_kept': len(optimized) / len(points),
                    'average_error': error_metrics['average_error'],
                    'max_error': error_metrics['max_error'],
                    'smoothness': calculate_smoothness(optimized)
                }
    
    return results

def plot_sensitivity_results(results):
    """
    Visualize the parameter sensitivity analysis results
    """
    plt.figure(figsize=(15, 10))
    
    # Create subplots for different metrics
    metrics = ['points_kept', 'average_error', 'max_error', 'smoothness']
    for i, metric in enumerate(metrics, 1):
        plt.subplot(2, 2, i)
        
        # Group results by method
        for method in ['stats', 'adaptive', 'percentile']:
            x_values = []
            y_values = []
            
            for key, value in results.items():
                if key[0] == method:  # key is (method, target, reg)
                    x_values.append(key[1])  # target_reduction
                    y_values.append(value[metric])
            
            plt.plot(x_values, y_values, 'o-', label=method)
        
        plt.xlabel('Target Reduction')
        plt.ylabel(metric.replace('_', ' ').title())
        plt.legend()
        plt.grid(True)
    
    plt.tight_layout()
    plt.show()

# Example usage:
param_ranges = {
    'methods': ['stats', 'adaptive', 'percentile'],
    'target_reductions': [0.3, 0.5, 0.7],
    'reg_strength': [0.01, 0.1, 0.5]
}

results = analyze_parameter_sensitivity(points, param_ranges)
plot_sensitivity_results(results)

# Print detailed results
for key, value in results.items():
    method, target, reg = key
    print(f"\nMethod: {method}, Target: {target}, Reg: {reg}")
    for metric, score in value.items():
        print(f"  {metric}: {score:.4f}")
```
